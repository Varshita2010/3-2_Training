### Noisy Channel MOdel 
- correcting the incorrect words.
- keyword -> noisy channel -> translate/output.
- why we need it?
    - because to check and rectify the errors.
##### Application :-
    1. Spell Checker("Piza" -> "Pizza").
    2. Speech Recognisation("IceCream"-"I Scream").
    3. Machine Translator(french -> english).
- checks logic:-
    - Likelihood
    - Prior probability
## N-gram Language Model:-
- used to estimate the probabilty of a word based on the previous word i.e.,(n-1).
- types of models:-
    1. unigram: autosuggestion (N-1) --> "hi","hello".
    2. bigram: paired words (N-2) --> "Merry" shows "Christmas".
    3. trigram: based on the previous two word gives next word (N-3) --> "I Love" suggest "Biryani".
- It focus on prediction.
## Evaluation Model:-
- it will evaluate the n-gram language.wheather it works prefectly or not.
- it wil works under the metrics i.e., perplexity.
- lower perplexity with higher model.
- higher perplexity with lower model(confused).
## Smoothing Techniques:-
- distribute mass to unseen n-gram model to avoid zero probablities. or simply andels zero probabilities.
- if it find new word in n-gram return zero as for the new word.
- it has two types: [use this for not getting crashed]
    1. Laplace : ( count - 1)
    2. Add-K: (count - 0.5)

```python
import nltk
from nltk.corpus import words
from collections import Counter

word_counts = {
    "the": 5000,
    "tea": 2000,
    "team": 1000,
    "ten": 200
}

summ = sum(word_counts.values())
def edit_distance(wrong_input, w):
    #likelihood --> how likely that particaular word repeat.
    distance = nltk.edit_distance(w,wrong_input)
    if distance == 0: return 1.0
    return 1/(10**distance)

def prior_probability(w):
    #prior_probability --> repeated words in training data model
    return word_counts.get(w,0)/summ

def suggested(wrong_input):
    candidates = list(word_counts.keys())
    correct_word = None
    high_prob = -1
    for w in candidates:
        likelihood = edit_distance(wrong_input,w)
        prior = prior_probability(w)
        result = likelihood * prior
        if result > high_prob:
            high_prob = result
            correct_word = w
    return correct_word

wrong_input = "teama"
correction = suggested(wrong_input)
print(correction)
```
