### Manual

  - word tokenization
  - sentence tokenizaiton

```python
# word tokenization
text = "Hi Spoorthi. Iam Varshita"
text.split()
```

```python
# sentence tokenization
text = "Hi Spoorthi. Iam Varshita"
text.split('.')
```

```python
# character tokenization
text = "we are undergoing training"
tokens = list(text)
print(tokens)
```

```python
# Even/Odd Number
number = int(input("enter a number to check : "))
if number%2==0:
    print(f"given number {number} is even")
else:
    print(f"Given number {number} is odd")
```

```python
# Perfect number
number = int(input("enter a number to check: "))
sum = 0
for i in range(1,number):
    if number%i==0:
        sum += i
if sum==number:
    print(f"given number {number} is a Perfect number")
else:
    print(f"given number {number} is not a Perfect number")
```

```python
# palindrone
number = int(input("enter a number to check: "))
pal = int(str(number)[::-1])
print(pal)
if pal==number:
    print("Palindrone")
else:
    print("not a palindrone")
```

```python
# Prime numbers
number = int(input("enter a number to check: "))
if number<=1:
    print("Not a Prime Number")
else:
    for i in range(2,number):
        if number%i==0:
            print(f"number {number} is not a Prime Number")
            break
        else:
            print(f"number {number} is  a prime number")
            break
```

```python
# Reverse a string
string = str(input("Enter a stringg to reverse: "))
rev_string = str(string)[::-1]
print(f"Reverse of given {string} is",rev_string)
```

```python
# fibanocci series
def fibannoci (n):
    a,b=0,1
    seq=[]
    for i in range(n):
        seq.append(a)
        a,b=b,a+b
    return seq
print(fibannoci(10))
```

```python
# fibanocci series
def fibannoci (n):
    a,b=0,1
    for i in range(n):
        print(a)
        c=a+b
        a=b
        b=c      
print(fibannoci(10))
```

```python
# factorial
number = int(input("Enter a number to check: "))
fact = 1
for i in range(1,number+1):
    fact *= i
print(fact)
```

```python
# Armstrong number
num = int(input("Enter a number: "))
temp = num
digits = len(str(num))
sum = 0

while temp > 0:
    digit = temp % 10
    sum += digit ** digits
    temp //= 10

if sum == num:
    print(num, "is an Armstrong number")
else:
    print(num, "is not an Armstrong number")
```
