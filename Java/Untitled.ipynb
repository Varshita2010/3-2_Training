```python
import random
import time
import os
WIDTH = 20
HEIGHT = 10

snake = [(5, 5)]
direction = 'd'
food = (random.randint(1, WIDTH-2), random.randint(1, HEIGHT-2))
score = 0

def draw():
    for y in range(HEIGHT):
        for x in range(WIDTH):
            if (x, y) in snake:
                print("O", end="")
            elif (x, y) == food:
                print("X", end="")
            elif x == 0 or x == WIDTH-1 or y == 0 or y == HEIGHT-1:
                print("#", end="")
            else:
                print(" ", end="")
        print()
    print("Score:", score)

draw()
```
