```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Lunch1.csv")
data
```

```python
data.info
```

```python
#data attributes
print(type(data))
print("data shape",data.shape)
print("data size",data.size)
```

```python
#printing all duplicated rows
data[data.duplicated(keep=False)]
```

```python
#drop duplicated rows
data.drop_duplicates(keep='first', inplace = True)
data
```

```python
#rename the columns
#CHANGE COLUMN NAME(RENAME THE COLUMNS)
data.rename({'Employee_ID': 'ID','Employee_Name':'Name','Lunch_Start_Time':'Start','Lunch_End_Time':'End','Lunch_Duration_Minutes':'Duration','Lunch_Location':'Location'}, axis=1, inplace = True)
data
```

```python
#drop duplicate column(temp c) and unamed column 
data1 = data.drop(["Meal_Type","Satisfaction_Rating"], axis=1)
data1
```

```python
#imput the missing values
#display the data1 info()
data1.info()
```

```python
#display data1 missing values count in each column using isnull().sum()
data1.isnull().sum()
```

```python
#viusalize dat1 missing values using 
cols = data1.columns
colors = ['black', 'blue']
sns.heatmap(data1[cols].isnull(),cmap=sns.color_palette(colors),cbar = True)
```

## Observation of the above graph
- the blue indicates the missing values
- the black represents the data without missing values
- the above graph tells us that there are no missing values in the data and the data puerly clean

```python
#find the mean and median values of each
median_Duration = data1["Duration"].median()
mean_Duration = data1["Duration"].mean()
print("Median of Duration: ", median_Duration)
print("Mean of Duration: ", mean_Duration)
```

## Why mean and median is used?
- mean and median values are used to train the model to draw the graphs
- and also to understand the overall tendency of the data
- Mean is used to understand the overall level(average value of the total data
- main reason is to fill the missing numeric values inn the data by the hellp of the mean and median values

```python
#replace the ozone missing values with median value
data1["Duration"] = data1['Duration'].fillna(median_Duration)
data1.isnull().sum()
```

```python
#find the mode values of categorcial column(weather)
print(data1["Start"].value_counts())
mode_start = data1["Start"].mode()[0]
print(mode_start)
```

```python
#impute missing values(replace nan with mode etc.) of "weather" using mh=fillnan()
data1["Start"] = data1["Start"].fillna(mode_start)
data1.isnull().sum()
```

```python
#find the mode values of categorcial column(weather)
print(data1["Department"].value_counts())
mode_dept = data1["Department"].mode()[0]
print(mode_dept)
```

```python
#impute missing values(replace nan with mode etc.) of "weather" using mh=fillnan()
data1["Department"] = data1["Department"].fillna(mode_dept)
data1.isnull().sum()
```

```python
#find the mode values of categorcial column(weather)
print(data1["End"].value_counts())
mode_end= data1["End"].mode()[0]
print(mode_dept)
```

```python
#impute missing values(replace nan with mode etc.) of "weather" using mh=fillnan()
data1["End"] = data1["End"].fillna(mode_end)
data1.isnull().sum()
```

```python
#find the mode values of categorcial column(weather)
print(data1["Location"].value_counts())
mode_loc = data1["Location"].mode()[0]
print(mode_loc)
```

```python
#impute missing values(replace nan with mode etc.) of "weather" using mh=fillnan()
data1["Location"] = data1["Location"].fillna(mode_loc)
data1.isnull().sum()
```

```python
#create a figure with two subplots, stacked vertically
fig, axes = plt.subplots(2, 1, figsize=(8,6),gridspec_kw={'height_ratios':[1,3]})
#plot the boxplot un the first (top) subplot
sns.boxplot(data=data1['Duration'], ax=axes[0], color='orange',width=0.5, orient = 'h')
axes[0].set_title("Boxplot")
axes[0].set_xlabel("Duration")

sns.histplot(data1["Duration"], kde=True, ax=axes[1], color='lavender', bins=30)
axes[1].set_title("Histogram with KDE")
axes[1].set_xlabel("Duration")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```
