## Excepiton Handling in java
- and disturbes the mormal flow of the program
- exeption = run time error that stops your program
> What is excception handling in java?
- is a mechanism in java to detect errors, handle errors, prevent program crash and continue normal execution.
- java uses 5 key words for exception handling.

### keywords
    keyword                 Purpose
    - try                   code that may cause exception
    - catch                 Handles the exception
    - finally               Always executes
    - throw                 Manually throw execption
    - throws                Declare exception in method

### types of exceptions in java
1. Checkd exception:
   - checked at compile time
   - must be handled using try catch and throw
   - ex:: io exceptiom and sql exceptions
2. Unchecked exceptions:
   - occur at run time
   - not compulsory to handle
   - ex:: nullpoint exception and arthimatic
3. Error
   - serious exceptions
   - cannot handled normally
   - ex:: out-of-memory error
